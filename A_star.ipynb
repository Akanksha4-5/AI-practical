```python
# A* Algorithm for 8 Puzzle Problem

class Node:

    def __init__(self, data, level, fval):

        self.data = data
        self.level = level
        self.fval = fval

    # Generate child nodes
    def generate_child(self):

        x, y = self.find(self.data, '_')

        moves = [
            [x, y - 1],
            [x, y + 1],
            [x - 1, y],
            [x + 1, y]
        ]

        children = []

        for i in moves:

            child = self.shuffle(self.data, x, y, i[0], i[1])

            if child is not None:
                children.append(Node(child, self.level + 1, 0))

        return children

    # Move blank space
    def shuffle(self, puz, x1, y1, x2, y2):

        if x2 >= 0 and x2 < 3 and y2 >= 0 and y2 < 3:

            temp = [row[:] for row in puz]

            temp[x1][y1], temp[x2][y2] = temp[x2][y2], temp[x1][y1]

            return temp

        return None

    # Find blank space
    def find(self, puz, x):

        for i in range(3):

            for j in range(3):

                if puz[i][j] == x:
                    return i, j


class Puzzle:

    def __init__(self):

        self.open = []
        self.closed = []

    # Heuristic function
    def h(self, start, goal):

        count = 0

        for i in range(3):

            for j in range(3):

                if start[i][j] != goal[i][j] and start[i][j] != '_':
                    count += 1

        return count

    # f(x) = h(x) + g(x)
    def f(self, start, goal):

        return self.h(start.data, goal) + start.level

    def process(self):

        start = [
            ['1', '2', '3'],
            ['_', '4', '6'],
            ['7', '5', '8']
        ]

        goal = [
            ['1', '2', '3'],
            ['4', '5', '6'],
            ['7', '8', '_']
        ]

        start = Node(start, 0, 0)

        start.fval = self.f(start, goal)

        self.open.append(start)

        while True:

            cur = self.open[0]

            print("\nCurrent State:\n")

            for i in cur.data:

                for j in i:
                    print(j, end=" ")

                print()

            # Goal state reached
            if self.h(cur.data, goal) == 0:
                print("\nGoal State Reached")
                break

            # Generate children
            for i in cur.generate_child():

                i.fval = self.f(i, goal)

                self.open.append(i)

            self.closed.append(cur)

            del self.open[0]

            # Sort by f value
            self.open.sort(key=lambda x: x.fval)


# Driver Code
puz = Puzzle()
puz.process()
```
