```python
import heapq

def prim(graph, start):
    visited = set()
    pq = [(0, start)]
    total_cost = 0
    
    while pq:
        cost, node = heapq.heappop(pq)
        
        if node not in visited:
            visited.add(node)
            total_cost += cost
            
            for neighbour, weight in graph[node]:
                if neighbour not in visited:
                    heapq.heappush(pq, (weight, neighbour))
    return total_cost
 

graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('A', 1), ('C', 2), ('D', 5)],
    'C': [('A', 4), ('B', 2), ('D', 1)],
    'D': [('B', 5), ('C', 1)]
}
 
print("total cost:" , prim(graph, 'A'))  
```
