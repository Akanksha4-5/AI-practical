```python
import random

# chatbot replies
answers = {
    "hi": [
        "Hello!",
        "Hi there!",
        "Hi! How can I help you?"
    ],

    "bye": [
        "Goodbye!",
        "See you later!",
        "Bye! Take care."
    ],

    "default": [
        "Sorry, I didn't understand that.",
        "Can you please rephrase?"
    ]
}

# function for bot reply
def chatbot(text):

    if text in answers:
        return random.choice(answers[text])

    else:
        return random.choice(answers["default"])


# starting message
print("Bot: Hi! Type 'quit' to stop chat.")

# chat loop
while True:

    user_text = input("You: ").lower()

    if user_text == "quit":
        print("Bot: Bye! Have a nice day.")
        break

print("Bot:", chatbot(user_text))
```
