```python
# N Queen Problem using Backtracking

N = 4


def printBoard(board):

    for row in board:
        print(row)


def isSafe(board, row, col):

    # Check left side row
    for i in range(col):

        if board[row][i] == 1:
            return False

    # Check upper diagonal
    i = row
    j = col

    while i >= 0 and j >= 0:

        if board[i][j] == 1:
            return False

        i -= 1
        j -= 1

    # Check lower diagonal
    i = row
    j = col

    while i < N and j >= 0:

        if board[i][j] == 1:
            return False

        i += 1
        j -= 1

    return True


def solve(board, col):

    # All queens placed
    if col >= N:
        return True

    for i in range(N):

        if isSafe(board, i, col):

            # Place queen
            board[i][col] = 1

            # Recursive call
            if solve(board, col + 1):
                return True

            # Backtracking
            board[i][col] = 0

    return False


# Main Program
board = [[0] * N for i in range(N)]

if solve(board, 0):

    print("Solution:")
    printBoard(board)

else:
    print("No Solution")
```
