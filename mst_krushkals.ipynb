```python
# Kruskal's Algorithm

class DS:

    def __init__(self, v):

        self.parent = {}

        for i in v:
            self.parent[i] = i

    # Find parent
    def find(self, x):

        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])

        return self.parent[x]

    # Union sets
    def union(self, a, b):

        self.parent[self.find(a)] = self.find(b)


def kruskal(vertices, edges):

    # Sort edges by weight
    edges.sort(key=lambda x: x[2])

    ds = DS(vertices)

    mst = []
    total = 0

    for u, v, w in edges:

        # Avoid cycle
        if ds.find(u) != ds.find(v):

            ds.union(u, v)

            mst.append((u, v, w))
            total += w

            print("Selected Edge:", (u, v, w))

    return mst, total


# Graph
vertices = ['A', 'B', 'C', 'D', 'E', 'F']

edges = [
    ('A', 'B', 11),
    ('A', 'C', 4),
    ('A', 'D', 8),
    ('B', 'E', 9),
    ('B', 'C', 1),
    ('D', 'C', 15),
    ('D', 'E', 2),
    ('C', 'F', 3),
    ('D', 'F', 6),
    ('F', 'E', 5)
]

# Function Call
mst, weight = kruskal(vertices, edges)

print("\nMinimum Spanning Tree:", mst)
print("Total Weight:", weight)
```
